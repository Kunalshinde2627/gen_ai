```python
import json

records = [
    {
        "id": 1,
        "name": "Rahul",
        "annual_income": 300000,
        "tax_paid": 0
    },
    {
        "id": 2,
        "name": "Amit",
        "annual_income": 400000,
        "tax_paid": 0
    },
    {
        "id": 3,
        "name": "Priya",
        "annual_income": 500000,
        "tax_paid": 0
    },
    {
        "id": 4,
        "name": "Sneha",
        "annual_income": 600000,
        "tax_paid": 0
    },
    {
        "id": 5,
        "name": "Akash",
        "annual_income": 700000,
        "tax_paid": 0
    }
]

with open("sample_records.json", "w") as file:
    json.dump(records, file, indent=4)

print("JSON file created successfully!")
```

```python
import json

with open("sample_records.json", "r") as file:
    data = json.load(file)

print(data)
```

```python
def display_records():
    for record in data:
        print("ID:", record["id"])
        print("Name:", record["name"])
        print("Annual Income:", record["annual_income"])
        print("Tax Paid:", record["tax_paid"])
        print("------------------------")


display_records()
```

```python
def search_record(name):
    for record in data:
        if record["name"].lower() == name.lower():
            print("Record Found!")
            print("ID:", record["id"])
            print("Name:", record["name"])
            print("Annual Income:", record["annual_income"])
            print("Tax Paid:", record["tax_paid"])
            return

    print("Record Not Found!")


search_record("Rahul")
```

```python
name = input("Enter name to search: ")

search_record(name)
```
